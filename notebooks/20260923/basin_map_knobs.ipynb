# Basin-definition map: knobs

`idd_tc_mortality.viz.basin_map.build_basin_map` on the idd-figures map stack. Load the land layer once
(cell 2), then edit the knobs cell and re-run it alone. All heights are inches; the map height follows the
extent aspect and the figure height is derived, so nothing is left to auto-layout. CLI equivalent: `run-basin-map`.

```python
%matplotlib inline
import geopandas as gpd
import matplotlib.pyplot as plt

from idd_figures.lib.io import save_figure
from idd_tc_mortality.viz import basin_map as bm

land = gpd.read_file(bm.ADMIN0_SIMPLIFIED)   # simplified LBD admin-0, offline
```

## Knobs

```python
knobs = dict(
    fig_width=16.0,
    central_longitude=180.0,        # Pacific-centred; 0.0 for Atlantic-centred
    title=bm.TITLE,
    title_fontsize=13,
    title_h=0.5,                    # height of the title band
    title_gap_h=0.35,               # gap between title band and map
    legend_gap_h=0.35,              # gap between map and legend
    legend_h=0.9,                   # height of the legend row
    legend_ncol=7,
    legend_fontsize=9,
    side_margin=0.02,               # left/right margin, figure fraction
    gridlines=True,                 # lat/lon gridlines with edge labels
    # painter knobs (basin_map_painter)
    water_color="#dbeaf4",
    land_color="#d8d5cc",
    border_color="white",
    label_fontsizes=(10, 8),        # (northern-hemisphere codes, southern-hemisphere two-line labels)
)
fig = bm.build_basin_map(land, **knobs)
```

Box coordinates, colours and label positions live in the module (`bm.IBTRACS_BOXES`, `bm.TCRISK_BOXES`,
`bm.NH_LABELS`, `bm.SH_LABELS`, `bm.BASIN_LEGEND`); edit them in place before the knobs cell if needed, e.g.
`bm.SH_LABELS[1] = (57, -40, "SI\\ntc-risk", "#D85A30")`.

## Save (working slot; overwritten on every run)

```python
save_figure(fig, bm.OUT_DIR / "basin_comparison_global.png", formats=["png", "pdf"], dpi=150)
```
